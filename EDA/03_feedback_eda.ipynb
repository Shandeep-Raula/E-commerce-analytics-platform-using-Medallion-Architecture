```python
import pandas as pd
import matplotlib.pyplot as plt

feedback_df = pd.read_csv("Feedback_KPI_Summary.csv")
sentiment_df = pd.read_csv("Sentiment_Category_Distribution.csv")
```

## Daily Total Reviews

```python
feedback_df["ORDER_DATE"] = pd.to_datetime(feedback_df["ORDER_DATE"])

monthly_reviews = (
    feedback_df
    .groupby(feedback_df["ORDER_DATE"].dt.month)["TOTAL_REVIEWS"]
    .sum()
)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_reviews.index,
    monthly_reviews.values,
    marker='o'
)

plt.title("Monthly Total Reviews")
plt.xlabel("Month Number")
plt.ylabel("Reviews")
plt.xticks(range(1,13))
plt.grid(alpha=0.3)

plt.show()
```

## Sentiment Distribution Analysis
Total Reviews by Sentiment

```python
plt.figure(figsize=(7,5))

plt.bar(
    sentiment_df["SENTIMENT_CATEGORY"],
    sentiment_df["TOTAL_REVIEWS"]
)

plt.title("Reviews by Sentiment Category")
plt.show()
```

## Percentage Share of Sentiments

```python
plt.figure(figsize=(7,7))

plt.pie(
    sentiment_df["PERCENT_SHARE"],
    labels=sentiment_df["SENTIMENT_CATEGORY"],
    autopct="%1.1f%%"
)

plt.title("Sentiment Distribution")
plt.show()
```

## Average Customer Rating Trend

```python
feedback_df["ORDER_DATE"] = pd.to_datetime(feedback_df["ORDER_DATE"])

monthly_reviews = (
    feedback_df
    .groupby(feedback_df["ORDER_DATE"].dt.month)["AVG_CUSTOMER_RATING"]
    .mean()
)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_reviews.index,
    monthly_reviews.values,
    marker='o'
)

plt.title("Monthly Total Reviews")
plt.xlabel("Month Number")
plt.ylabel("Reviews")
plt.xticks(range(1,13))
plt.grid(alpha=0.3)

plt.show()
```
