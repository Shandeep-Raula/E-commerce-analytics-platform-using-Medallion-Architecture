```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
channel_df = pd.read_csv("Session_Channel_Performance_View.csv")
customer_df = pd.read_csv("Session_Customer_Segmentation_View.csv")
funnel_df = pd.read_csv("Session_Funnel_Analysis_View.csv")
kpi_df = pd.read_csv("Session_KPI_Aggregation_View.csv")
```

## Channel Performance Analysis
Total Sessions by Channel

```python
plt.figure(figsize=(10,5))

plt.bar(
    channel_df["CHANNEL_NAME"],
    channel_df["CONVERSION_RATE"],
    color=[
        "blue", "green", "orange", "red", "purple",
        "cyan", "darkgreen", "gold", "tomato", "darkmagenta"
    ]
)

plt.title("Conversion Rate by Channel")
plt.xticks(rotation=45)

plt.show()
```

Conversion Rate by Channel

```python
plt.figure(figsize=(10,5))

plt.bar(
    channel_df["CHANNEL_NAME"],
    channel_df["CONVERSION_RATE"],
    color=[
        "blue", "green", "orange", "red", "purple",
        "cyan", "darkgreen", "gold", "tomato", "darkmagenta"
    ])

plt.title("Conversion Rate by Channel")
plt.xlabel("Channel")
plt.ylabel("Conversion Rate")
plt.xticks(rotation=45)

plt.show()
```

Average Session Duration

```python
plt.figure(figsize=(10,5))

plt.bar(
    channel_df["CHANNEL_NAME"],
    channel_df["AVG_SESSION_DURATION"],
    color=[
        "blue", "green", "orange", "red", "purple",
        "cyan", "darkgreen", "gold", "tomato", "darkmagenta"
    ]
)

plt.title("Average Session Duration")
plt.xticks(rotation=45)
plt.show()
```

## Customer Segmentation Analysis

```python
customer_df.groupby(
    "AGE_GROUP"
)["TOTAL_CUSTOMERS"].sum()
```

```python
customer_age = customer_df.groupby(
    "AGE_GROUP"
)["TOTAL_CUSTOMERS"].sum()

plt.figure(figsize=(8,5))

plt.bar(
    customer_age.index,
    customer_age.values,
    color=["skyblue", "orange", "green", "purple", "coral"]
)

plt.title("Customers by Age Group")
plt.xticks(rotation=20)
plt.show()
```

## Purchases by Income Bracket

```python
income_purchase = customer_df.groupby(
    "INCOME_BRACKET"
)["TOTAL_PURCHASES"].sum()

plt.figure(figsize=(8,5))

plt.bar(
    income_purchase.index,
    income_purchase.values,
    color=["skyblue", "orange", "green", "purple", "coral"]
)

plt.title("Purchases by Income")
plt.xticks(rotation=30)
plt.show()
```

## Conversion Rate by Gender

```python
customer_df.groupby(
    "GENDER"
)["CONVERSION_RATE"].mean()
```

## Funnel Analysis
Product Views vs Purchases

```python
plt.figure(figsize=(12,5))

plt.plot(
    funnel_df["MONTH_NAME"],
    funnel_df["PRODUCT_VIEWS"],
    marker='o'
)

plt.plot(
    funnel_df["MONTH_NAME"],
    funnel_df["PURCHASES"],
    marker='o'
)

plt.legend(["Views","Purchases"])

plt.title("Views vs Purchases")
plt.show()
```

## View to Cart Rate

```python
plt.figure(figsize=(10,5))

plt.bar(
    funnel_df["MONTH_NAME"],
    funnel_df["VIEW_TO_CART_RATE"]
)

plt.title("View to Cart Rate")
plt.show()
```

## Checkout to Purchase Rate

```python
plt.figure(figsize=(10,5))

plt.bar(
    funnel_df["MONTH_NAME"],
    funnel_df["CHECKOUT_TO_PURCHASE_RATE"]
)

plt.title("Checkout to Purchase Rate")
plt.show()
```
